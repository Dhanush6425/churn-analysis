Hypothesis Testing

```python

def hypothesis_result(p_value, alpha = 0.05):

    if p_value < alpha:

        print("Reject H0")
        return 1

    else:

        print("Failed to reject H0")
        return 0
 
```

### Chi Square Test

```python
from scipy.stats import chi2_contingency

def chi_square(columns):
    table = pd.crosstab(df[columns],df['churn_label'])
    chi2,p,dof,expected = chi2_contingency(table)
    print("p-value: ",p)
    print("Result for Hypothesis: ",end='')
    a = hypothesis_result(p)
    return a
```

###T Test

```python
from scipy.stats import ttest_ind

def t_test(columns):
    group_0 = df[df['churn_label'] == 0][columns]
    group_1 = df[df['churn_label'] == 1][columns]
    t_stat,p=ttest_ind(group_0,group_1)
    print("p-value: ",p)
    print("Result for Hypothesis: ",end='')
    a = hypothesis_result(p)
    return a
```
